| Pieza | ¿Qué hace? | Analogía |

| `float(...)` | Convierte el texto escrito en un número con decimales. | Tu orden está escrita en papel; `float()` la convierte en una cantidad con la que la cocina puede trabajar. |

| `print(f"...")` | Muestra un mensaje en pantalla. Lo que va entre `{ }` se reemplaza por el valor de la variable. | El ticket que te entregan al final con tu cuenta. |

## Paso 1 — Ejecuta la calculadora (versión de 3 plataformas)

Ejecuta la celda de abajo con `Shift + Enter`. El programa te hará preguntas en un cuadro de texto: escribe tu respuesta y presiona `Enter` después de cada una. Usa **tus horas reales**. Para escribir media hora usa punto, no coma: `1.5`.

```python
nombre = input("¿Cómo te llamas? ")
horas_instagram = float(input("Horas al día en redes sociales(instagram): "))
horas_Whatsapp = float(input("Horas al día en mensajería: "))
horas_Tik_Tok= float(input("Horas al día en streaming (series, videos): "))
horas_transporte = float(input("Horas al dia en transporte(uber)"))
horas_youtube_music = (float(input("Horas al dia en musica")))

tiempo_total = horas_instagram + horas_Whatsapp + horas_Tik_Tok + horas_transporte + horas_youtube_music
porcentaje = (tiempo_total / 24) * 100

print("----- RESUMEN DE TU RUTINA DIGITAL -----")
print(f"Usuario: {nombre}")
print(f"Tiempo total en pantalla: {tiempo_total} horas")
print(f"Porcentaje de tu día: {porcentaje:.2f}%")
```
